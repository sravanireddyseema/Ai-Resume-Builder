```python
import pandas as pd

df = pd.read_excel("ApexPlanet_DataAnalytics_Dataset.xlsx")

print(df.head())
print(df.info())

# Check missing values
print("\nMissing values:")
print(df.isnull().sum())

# Check duplicates
print("\nDuplicates:")
print(df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()

# Standardize City
df["City"] = df["City"].str.strip().str.title()

# Check again after cleaning
print("\nMissing values after cleaning:")
print(df.isnull().sum())

print("\nData types:")
print(df.dtypes)

print("\nDuplicates after cleaning:")
print(df.duplicated().sum())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Order_Year"] = df["Order_Date"].dt.year
```

```python
df["Order_Month"] = df["Order_Date"].dt.month
```

```python
df["Order_Month_Name"] = df["Order_Date"].dt.month_name()
```

```python
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[17, 25, 35, 45, 55, 65],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65"]
)
```

```python
df["Sales_Category"] = pd.cut(
    df["Total_Sales"],
    bins=[0, 50000, 150000, float("inf")],
    labels=["Low", "Medium", "High"]
)
```

```python
print("\nOrder Year:")
print(df["Order_Year"].value_counts())

print("\nOrder Month:")
print(df["Order_Month"].value_counts().sort_index())

print("\nAge Groups:")
print(df["Age_Group"].value_counts().sort_index())

print("\nSales Categories:")
print(df["Sales_Category"].value_counts())
```
